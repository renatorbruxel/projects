# Inspect the shared Kaggle CRM data

Source: [CRM Sales Opportunities](https://www.kaggle.com/datasets/nilkamalsaha/crm-sales-opportunities-on-google-sheets), version 1, CC0.

Run from the repository or extracted package. No external packages or network calls are required. Original years and missing values are preserved. The build validates raw hashes and writes deterministic clean files.


```python
from pathlib import Path
import runpy, json
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p/'scripts/build_crm_dataset.py').exists())
module = runpy.run_path(str(root/'scripts/build_crm_dataset.py'))
data = module['build']()
print(json.dumps(data['summary'], indent=2))
```

```python
q = data['quality']
assert q['rows_before_join'] == q['rows_after_join'] == 8800
assert q['duplicate_opportunity_ids'] == 0
assert q['won_value_reconciles_to_raw']
assert q['open_close_values_remain_null']
print(json.dumps(q, indent=2))
```

## Interpretation

1,425 account references remain missing; account-level analyses must retain an Unknown group. Open opportunities have no deal value, so monetary pipeline coverage cannot be derived. The GTXPro alias affects 1,480 rows and is repaired before joining. Five of the 35 registered sellers have no opportunities. First-observed purchases cannot establish genuine Net New status because earlier customer history is unavailable.

Dollar presentation assumes USD; account annual revenue is explicitly in USD millions, while the source dictionary does not separately name the deal-value currency. No quotas, CAC, churn, forecasts, stage histories or approvals are fabricated.
